# 01 — First Detection

Exploratory notebook for Phase 1. We run the hold detector on a sample image and inspect the result inline.

**Reminder:** the MVP uses generic `yolov8n` (COCO), so on a real wall photo it will detect few/zero holds. The goal here is to see the full pipeline run and to tune HSV color ranges. See `models/README.md` for training a hold-specific model.

```python
import os, sys

# Make src/ importable from the notebooks/ folder.
PROJECT_ROOT = os.path.abspath('..')
sys.path.insert(0, os.path.join(PROJECT_ROOT, 'src'))

import cv2
import matplotlib.pyplot as plt
import utils
import hold_detector
```

```python
# Point this at a climbing-wall image you've placed in data/input/.
IMAGE_PATH = os.path.join(PROJECT_ROOT, 'data', 'input', 'wall.jpg')

output_path = hold_detector.run(IMAGE_PATH)
print('Wrote:', output_path)
```

```python
# Show the annotated result. cv2 loads BGR; matplotlib expects RGB, so convert.
annotated = cv2.imread(output_path)
plt.figure(figsize=(12, 8))
plt.imshow(cv2.cvtColor(annotated, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.title('Detected holds')
plt.show()
```

## Tuning color ranges

If colors look wrong, edit the `colors:` block in `config/settings.yaml` and re-run. The cell below helps you inspect the HSV value of a specific pixel to calibrate ranges.

```python
img = utils.read_image(IMAGE_PATH)
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

# Inspect the HSV of a pixel (row y, col x) sitting on a hold of known color.
y, x = 100, 100
print('HSV at', (x, y), '=', hsv[y, x])
```
